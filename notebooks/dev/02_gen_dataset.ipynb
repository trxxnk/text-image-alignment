```python
import os, sys
os.chdir(os.path.dirname(sys.prefix))
```

```python
from src.tsp_dewarp.transforms import (
    Compose,
    IdentityTransform,
    RandomWaveTransform,
    RandomPerspectiveTransform
)
from src.tsp_dewarp.dataset import TPSDatasetGenerator
```

```python
identity_pipeline = Compose([
    IdentityTransform()
])
```

```python
perspective_pipeline = Compose([
    RandomPerspectiveTransform(
        distortion_scale_range=(0.05, 0.10)  # 5.0% - 10.0%
    )
])
```

```python
wave_pipeline = Compose([
    RandomWaveTransform(
        amp_x_range=(0.01, 0.03),   # 1.0% - 3.0%
        amp_y_range=(0.01, 0.03),
        freq_range=(0.5, 1.0)
    )
])
```

```python
combo_pipeline = Compose([
    RandomPerspectiveTransform(
        distortion_scale_range=(0.01, 0.04)
    ),
    RandomWaveTransform(
        amp_x_range=(0.01, 0.04),
        amp_y_range=(0.01, 0.04),
        freq_range=(0.6, 1.2)
    )
])
```

```python
transform_configs = [
    {"difficulty": "easy", "pipeline": perspective_pipeline, "prob": 0.4},
    {"difficulty": "medium", "pipeline": wave_pipeline, "prob": 0.2},
    {"difficulty": "hard", "pipeline": combo_pipeline, "prob": 0.2},
    {"difficulty": "identity", "pipeline": identity_pipeline, "prob": 0.2},
]
```

```python
generator = TPSDatasetGenerator(
    input_dir="./data/raw/v3/",           # исходные ровные страницы
    output_dir="./data/generated/v3/",    # новая версия датасета
    transform_configs=transform_configs,
    grid_size=9,                          # 9x9 TPS
    random_seed=42
)
```

```python
generator = TPSDatasetGenerator(
    input_dir="./data/raw/v1",           # исходные ровные страницы
    output_dir="./data/generated/test2/",    # новая версия датасета
    transform_configs=transform_configs,
    grid_size=9,                          # 9x9 TPS
    random_seed=42
)
```
