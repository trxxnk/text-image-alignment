```python
import json
import mlflow
from mlflow.client import MlflowClient
```

```python
import dagshub
dagshub.init(repo_owner='trxxnk', repo_name='text-image-alignment', mlflow=True)
```

```python
def export_run_to_json(run_id: str, output_file: str = "artifacts/mlflow_run_data.json"):
    # 1. Инициализация клиента MLflow
    client = MlflowClient()

    # 2. Получение базовой информации о запуске (параметры и теги)
    run_info = client.get_run(run_id)
    parameters = run_info.data.params
    tags = run_info.data.tags

    # 3. Получение списка всех доступных имен метрик для этого запуска
    metric_names = run_info.data.metrics.keys()

    # 4. Сбор полной истории по каждой метрике (по эпохам/шагам)
    metrics_history = {}
    for name in metric_names:
        history = client.get_metric_history(run_id, name)

        # Сохраняем в виде списка словарей со связкой: шаг -> значение
        metrics_history[name] = [
            {"step": m.step, "value": m.value, "timestamp": m.timestamp}
            for m in history
        ]

    # 5. Формирование единой структуры данных
    export_data = {
        "run_id": run_id,
        "experiment_id": run_info.info.experiment_id,
        "status": run_info.info.status,
        "parameters": parameters,
        "metrics_history": metrics_history,
        "tags": tags,
    }

    # 6. Запись в JSON-файл с отступами для читаемости
    with open(output_file, "w", encoding="utf-8") as f:
        json.dump(export_data, f, indent=4, ensure_ascii=False)

    print(f"Данные успешно выгружены в файл: {output_file}")
```

```python
RUN_ID = "955b9fee20d8409492c1554196a6d673"
export_run_to_json(run_id=RUN_ID, output_file="artifacts/mlflow_run_data.json")
```
